### Temuan Kualitas Data dan Keputusan Penanganan

Berdasarkan hasil pemeriksaan awal terhadap dataset kualitas udara Yogyakarta 2021, berikut adalah beberapa temuan utama beserta keputusan pembersihannya:

1. **Pemeriksaan Data Kosong (*Missing Values*):**
   * **Temuan:** Pada beberapa kolom parameter polutan (seperti `PM2.5`, `PM10`, dan `SO2`), terdapat nilai kosong/hilang pada tanggal atau jam tertentu akibat keterbatasan pencatatan sensor stasiun.
   * **Keputusan:** **Dibiarkan** untuk kolom yang data kosongnya sedikit, atau **dihapus barisnya** jika kolom utama yang dianalisis tidak memiliki angka sama sekali. 
   * **Alasan:** Menghapus baris yang memiliki data kosong secara wajar tidak mengurangi representasi data secara signifikan karena jumlah total baris masih jauh di atas kriteria minimal (200 baris).

2. **Kesesuaian Tipe Data (*Data Types*):**
   * **Temuan:** Kolom pengukuran polutan numerik (seperti `PM2.5`, `PM10`, `CO`, dll.) terbaca sebagai tipe data `float64` atau `int64`, yang sudah sesuai sebagai variabel kuantitatif. Tipe data kolom tanggal (`Date`) terbaca sebagai `object` (teks).
   * **Keputusan:** **Diperbaiki/Dibiarkan**. Tipe data numerik dibiarkan karena sudah siap dihitung secara matematis. Kolom `Date` dapat diubah ke tipe `datetime` jika diperlukan untuk analisis berbasis waktu.
   * **Alasan:** Memastikan semua nilai yang dihitung untuk rata-rata dan statistik deskriptif berada dalam format numerik yang valid.

3. **Pemeriksaan Nilai yang Tidak Masuk Akal (*Outliers / Irreasonable Values*):**
   * **Temuan:** Nilai konsentrasi polutan bernilai $0$ atau angka positif yang wajar (tidak ditemukan nilai negatif seperti $-5$ atau angka ekstrem yang tidak wajar).
   * **Keputusan:** **Dibiarkan**.
   * **Alasan:** Angka-angka tersebut masih berada pada rentang rasional pengukuran kualitas udara ambien.

4. **Pemeriksaan Baris Duplikat (*Duplicate Rows*):**
   * **Temuan:** Tidak ditemukan adanya baris duplikat ganda (`0` duplikat) pada dataset.
   * **Keputusan:** **Dibiarkan / Tidak perlu tindakan**.
   * **Alasan:** Setiap baris mewakili data waktu pencatatan unik yang tidak berulang secara tidak sengaja.

```python
import pandas as pd
import numpy as np

kolom_numerik = ["PM2.5", "PM10", "CO"]

def statistik_deskriptif(df, kolom_list):
    ringkasan = {}
    for col in kolom_list:
        data = df[col].dropna()
        q1 = data.quantile(0.25)
        q3 = data.quantile(0.75)
        iqr = q3 - q1
        
        mode_val = data.mode()[0] if not data.mode().empty else np.nan
        
        ringkasan[col] = {
            "Mean (Rata-rata)": data.mean(),
            "Median": data.median(),
            "Modus": mode_val,
            "Minimum": data.min(),
            "Maksimum": data.max(),
            "Jangkauan (Range)": data.max() - data.min(),
            "Varians": data.var(),
            "Simpangan Baku (Std Dev)": data.std(),
            "Q1 (Kuartil 1)": q1,
            "Q3 (Kuartil 3)": q3,
            "IQR": iqr
        }
    return pd.DataFrame(ringkasan)

tabel_ringkasan = statistik_deskriptif(df, kolom_numerik)
display(tabel_ringkasan.round(2))

print("\n=== NILAI PLUS: Perhitungan Manual Varians 10 Baris Pertama (PM2.5) ===")
sample_10 = df["PM2.5"].dropna().head(10).tolist()

n = len(sample_10)
mean_manual = sum(sample_10) / n
deviasi_kuadrat = [(x - mean_manual)**2 for x in sample_10]
var_manual = sum(deviasi_kuadrat) / (n - 1)
std_manual = math.sqrt(var_manual) if 'math' in globals() else np.sqrt(var_manual)

print(f"Data 10 baris pertama : {sample_10}")
print(f"Rata-rata (mean)     : {mean_manual:.2f}")
print(f"Varians Manual (n-1)  : {var_manual:.2f}")
print(f"Varians Pandas       : {df['PM2.5'].dropna().head(10).var():.2f}")
print(f"Std Deviasi Manual   : {std_manual:.2f}")
print(f"Std Deviasi Pandas   : {df['PM2.5'].dropna().head(10).std():.2f}")
```
